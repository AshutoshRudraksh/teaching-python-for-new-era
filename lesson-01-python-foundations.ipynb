## 2. Variables: giving names to values

A variable is a name that holds a value, like a labelled box. We write the name, then `=`, then the value.

```python
patient_name = "Ravi"
patient_age = 42

print(patient_name)
print(patient_age)
```

Notice two things:
- Text (called a **string**) goes inside quotes.
- Numbers do not need quotes.

## 3. Numbers and simple maths

Python is a calculator that remembers. The symbols are `+`, `-`, `*` (multiply) and `/` (divide).

```python
visits_monday = 12
visits_tuesday = 9
visits_wednesday = 15

total = visits_monday + visits_tuesday + visits_wednesday
print("Total visits:", total)
print("Average per day:", total / 3)
```

## 4. Lists: many values under one name

A **list** holds many values in order, inside square brackets. Two helpers you will use every week:
- `sum(...)` adds up a list of numbers
- `len(...)` counts how many items are in a list

Together they give you an average: sum divided by count.

```python
patient_ages = [42, 35, 58, 29, 64]

total_age = sum(patient_ages)
count = len(patient_ages)
average_age = total_age / count

print("Total age:", total_age)
print("Number of patients:", count)
print("Average age:", average_age)
```

**Your turn:** add two more ages to the list, predict the new average roughly in your head, then run and check.

```python
patient_ages = [42, 35, 58, 29, 64]  # add two ages inside the brackets

print("Average age:", sum(patient_ages) / len(patient_ages))
```

## 5. Looking at one item at a time: the `for` loop

A loop repeats steps for every item in a list. Read this as: "for each age in patient_ages, print the age." 

```python
for age in patient_ages:
    print(age)
```

## 6. Making decisions: `if`

Add an `if` inside the loop and the computer can act only on some items. Here we print only the patients who are 50 or older.

```python
for age in patient_ages:
    if age >= 50:
        print(age, "is 50 or older")
```

## 7. Records: a small table in code

Real work data comes as records: each record has named fields. In Python we write one record as a **dictionary** with keys like `"name"`, `"age"` and `"condition"`, and a list of records works like a small spreadsheet.

```python
patients = [
    {"name": "Ravi",  "age": 42, "condition": "Diabetes"},
    {"name": "Meena", "age": 35, "condition": "Checkup"},
    {"name": "Arun",  "age": 58, "condition": "Diabetes"},
    {"name": "Sara",  "age": 29, "condition": "Checkup"},
    {"name": "Kiran", "age": 64, "condition": "Blood pressure"},
]

for patient in patients:
    print(patient["name"], patient["age"], patient["condition"])
```

## 8. Counting by category

This is the pattern behind half of all weekly reports: go through every record, and keep a running count for each category.

```python
visits_by_condition = {}

for patient in patients:
    condition = patient["condition"]
    if condition in visits_by_condition:
        visits_by_condition[condition] = visits_by_condition[condition] + 1
    else:
        visits_by_condition[condition] = 1

print(visits_by_condition)
```
