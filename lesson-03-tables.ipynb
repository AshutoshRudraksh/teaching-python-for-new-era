## The one new idea: a table is a list of rows

Look at any spreadsheet you use. It has **rows** going down and **columns** going across.

In Python, the cleanest way to hold that is:

- The whole table is a **list**.
- Each **row** is a **dictionary**: the column names are the keys, that row's values are the values.
- A **column** is one key, read down every row.

Here is one week at the clinic. Five rows, one per day. Read it like a spreadsheet first: what are the columns? What does one row describe?

```python
week = [
    {"day": "Mon", "checkups": 12, "diabetes": 5, "blood_pressure": 4, "avg_wait": 18},
    {"day": "Tue", "checkups": 10, "diabetes": 6, "blood_pressure": 3, "avg_wait": 22},
    {"day": "Wed", "checkups": 14, "diabetes": 4, "blood_pressure": 5, "avg_wait": 16},
    {"day": "Thu", "checkups": 11, "diabetes": 7, "blood_pressure": 4, "avg_wait": 20},
    {"day": "Fri", "checkups": 13, "diabetes": 6, "blood_pressure": 6, "avg_wait": 19},
]

print("Rows in the table:", len(week))
print("First row:", week[0])
```

## Working one row, then every row

A function that understands **one row** is easy to write and easy to check. Once it is right for one row, a loop lets it work for every row. That is the pattern for this whole lesson: **one row first, then the table.**

Total visits on one day is just the three visit columns added together. Predict Monday's total before you run the next cell.

```python
def day_total(row):
    return row["checkups"] + row["diabetes"] + row["blood_pressure"]

print("Monday total:", day_total(week[0]))

for row in week:
    print(row["day"], day_total(row))
```

## Reading a whole column

A column total asks the opposite question: one column, every row. Same pattern as your Lesson 1 counting, pointed at a column instead.

**Predict first:** total checkups for the week, and total diabetes visits. Add them in your head or on paper before running.

```python
def column_total(table, column):
    total = 0
    for row in table:
        total = total + row[column]
    return total

print("Checkups this week:", column_total(week, "checkups"))
print("Diabetes this week:", column_total(week, "diabetes"))
print("Blood pressure this week:", column_total(week, "blood_pressure"))
```

Notice what just happened. `column_total` does not know or care which column you ask for. The column name is a **parameter**, so one function serves every column in the table. That is the payoff of Lesson 2: write it once, generally, and reuse it everywhere.

Two more questions a manager always asks: which day was busiest, and how long did patients wait on average? Both are table functions too.

```python
def busiest_day(table):
    best_day = None
    best_total = 0
    for row in table:
        if day_total(row) > best_total:
            best_total = day_total(row)
            best_day = row["day"]
    return best_day

def average_wait(table):
    waits = []
    for row in table:
        waits.append(row["avg_wait"])
    return sum(waits) / len(waits)

print("Busiest day:", busiest_day(week))
print("Average wait:", average_wait(week))
```

## Build the module: `table_tools`

Same move as Lesson 2. Functions you will use every week do not live scattered in a notebook; they live in a module. Save all four recipes into `table_tools.py`, import it, and the notebook becomes questions and answers only.

```python
%%writefile table_tools.py

def day_total(row):
    return row["checkups"] + row["diabetes"] + row["blood_pressure"]


def column_total(table, column):
    total = 0
    for row in table:
        total = total + row[column]
    return total


def busiest_day(table):
    best_day = None
    best_total = 0
    for row in table:
        if day_total(row) > best_total:
            best_total = day_total(row)
            best_day = row["day"]
    return best_day


def average_wait(table):
    waits = []
    for row in table:
        waits.append(row["avg_wait"])
    return sum(waits) / len(waits)
```

## The weekly summary, in one cell

**Predict the whole summary first** from the table at the top: week total, the three column totals, busiest day, average wait. Then run this cell and compare. Every number here comes from `table_tools`; this cell contains no arithmetic at all.

```python
import table_tools

week_total = 0
for row in week:
    week_total = week_total + table_tools.day_total(row)

print("Weekly summary")
print("--------------")
print("Total visits:", week_total)
print("Checkups:", table_tools.column_total(week, "checkups"))
print("Diabetes:", table_tools.column_total(week, "diabetes"))
print("Blood pressure:", table_tools.column_total(week, "blood_pressure"))
print("Busiest day:", table_tools.busiest_day(week))
print("Average wait:", table_tools.average_wait(week))
```
