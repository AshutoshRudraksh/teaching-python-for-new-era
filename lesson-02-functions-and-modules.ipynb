## Part A, the theory of functions

### A function is a named recipe

Think of a recipe card. The card has a **name**, a list of **ingredients** it needs, **steps**, and a **finished dish** it hands back.

In Python:

- `def` starts the recipe card. It means "define a function".
- The **name** is how you call it later.
- The names inside the brackets are **parameters**: placeholders for the ingredients.
- The values you pass when you call it are **arguments**: the actual ingredients that day.
- `return` hands the finished dish back to whoever called the function.

One card, many meals. You write the steps once; the ingredients can change every time.

````python
def average_age(ages):
    total = sum(ages)
    return total / len(ages)

# Call the recipe with today's ingredients
print(average_age([42, 35, 58, 29, 64]))
````

**Predict first.** The same recipe, new ingredients. What will these two lines print? Say it out loud before running.

````python
print(average_age([30, 40, 50]))
print(average_age([25, 25]))
````

### `return` is not `print`

This is the most important distinction in this lesson.

- `print` only **shows** a value on the screen. After that, the value is gone.
- `return` **gives the value back**, so the rest of your program can store it, compare it, or calculate with it.

Watch: because `average_age` returns its answer, we can save it in a variable and keep working with it.

````python
result = average_age([42, 35, 58, 29, 64])
print("Saved answer:", result)
print("Double the average age would be:", result * 2)
````

### Scope: what happens inside a function stays inside it

Variables created inside a function are **local**: they exist only while the function runs, then they disappear. That is a feature, not a bug. It means one function's scratch work can never corrupt another function's.

In the cell below, `total` exists only inside `average_age`. The last line would fail with a `NameError` if you uncommented it, because outside the function there is no `total`. Try it if you like, then comment it back.

````python
def average_visits(counts):
    total = sum(counts)          # 'total' is local, it lives only in here
    return total / len(counts)

print(average_visits([12, 9, 15]))

# print(total)   # This would fail: NameError. 'total' does not exist out here.
````

### Good functions follow three rules

1. **One job.** A function does one clear thing: one average, one count, one list.
2. **Data in, answer out.** It takes its data through parameters and hands the answer back with `return`. It does not reach outside itself for data.
3. **A name that says what it returns.** `average_age`, `count_visits`, `older_names`. If you can guess what a function does from its name, the name is right.

Keep these three rules and your code stays easy to trust, which matters when the numbers feed a real report.

## Part B, the theory of modules

### A module is a file full of recipes

A function makes one recipe reusable **inside one notebook**. A **module** makes recipes reusable **everywhere**.

A module is simply a Python file, ending in `.py`, that contains functions. Any notebook or program can then write:

```python
import clinic_tools
```

and borrow every function in that file, like this: `clinic_tools.average_age(...)`.

The part before the dot is the **namespace**: it says which toolbox the tool came from, so two modules can both have a function called `average` without ever colliding.

Python also ships with a toolbox of ready made modules called the **standard library**. Other people's tested functions, free to import. Example: the `statistics` module already knows how to compute a mean.

````python
import statistics

ages = [42, 35, 58, 29, 64]
print("Our function:      ", average_age(ages))
print("statistics.mean(): ", statistics.mean(ages))
````

Same answer, two sources. That is the whole idea: **never write what you can borrow, and write your own so you can borrow them forever.**

## Part C, practical walkthrough: build `clinic_tools` and run a weekly report

We will do this in four steps. Type each step yourself; do not copy and paste. Your fingers learn what your eyes skip.

**Step 1.** Create the module. The special first line, `%%writefile clinic_tools.py`, tells Colab: save everything in this cell as a file named `clinic_tools.py`. Inside it go the three recipes from your own work so far: average age, visits counted by condition, and the names of patients over an age limit.

````python
%%writefile clinic_tools.py

def average_age(ages):
    total = sum(ages)
    return total / len(ages)


def count_by_condition(records):
    counts = {}
    for record in records:
        condition = record["condition"]
        if condition in counts:
            counts[condition] = counts[condition] + 1
        else:
            counts[condition] = 1
    return counts


def older_names(records, age_limit):
    names = []
    for record in records:
        if record["age"] >= age_limit:
            names.append(record["name"])
    return names
````

**Step 2.** Import your module and meet your data. This is the same style of patient list as Lesson 1, with one extra patient.

**Predict first.** Before Step 3, work out by hand: the average age, the count for each condition, and who is 50 or older. The module must match your hand answers, or we debug together.

````python
import clinic_tools

patients = [
    {"name": "Asha",  "age": 34, "condition": "Checkup"},
    {"name": "Ravi",  "age": 58, "condition": "Diabetes"},
    {"name": "Meena", "age": 47, "condition": "Blood pressure"},
    {"name": "Arun",  "age": 66, "condition": "Diabetes"},
    {"name": "Kiran", "age": 52, "condition": "Checkup"},
]

ages = [patient["age"] for patient in patients]
print("Patients this week:", len(patients))
````

**Step 3.** Produce the weekly report, using only borrowed functions. Notice what is missing: no loops, no sums, no counting code in this cell at all. The work lives in the module; this cell just asks questions and prints answers. That separation is what makes a report you can rerun every week without fear.

````python
print("Weekly report")
print("-------------")
print("Average age:", clinic_tools.average_age(ages))
print("Visits by condition:", clinic_tools.count_by_condition(patients))
print("Patients 50 or older:", clinic_tools.older_names(patients, 50))
````

**Step 4.** New week, new data, zero new code. Add one patient, predict every line of the report again, then rerun Step 3 exactly as it is. If the module is right, only the data changes.

````python
patients.append({"name": "Sana", "age": 29, "condition": "Diabetes"})
ages = [patient["age"] for patient in patients]
print("Patients now:", len(patients))
````
