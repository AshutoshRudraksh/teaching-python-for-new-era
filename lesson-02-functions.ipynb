## 2. The one new idea: a function is a named recipe

In Lesson 1, calculating the average took two ideas: `sum` and `len`. A **function** lets you name those steps once, then reuse the name whenever you need them.

Read this slowly:
- `def` starts the recipe and gives it a name.
- The value in brackets, `ages`, is the **input**.
- `return` hands the answer back.
- Calling the function means writing its name with a value.

```python
def average_age(ages):
    total = sum(ages)
    return total / len(ages)

print(average_age([42, 35, 58, 29, 64]))
```

**Predict first:** run the next cell in your head. The same recipe, new numbers. That is the whole point of a recipe.

```python
print(average_age([30, 40, 50]))
print(average_age([25, 25]))
```

## 4. Turn Lesson 1 counting into a function

In Lesson 1 you counted visits by condition with a loop. That whole block is a recipe too. Here it is as a function: the records go in, the counts come back.

```python
def count_visits(records):
    counts = {}
    for record in records:
        condition = record["condition"]
        if condition in counts:
            counts[condition] = counts[condition] + 1
        else:
            counts[condition] = 1
    return counts

patients = [
    {"name": "Ravi",  "age": 42, "condition": "Diabetes"},
    {"name": "Meena", "age": 35, "condition": "Checkup"},
    {"name": "Arun",  "age": 58, "condition": "Diabetes"},
    {"name": "Sara",  "age": 29, "condition": "Checkup"},
    {"name": "Kiran", "age": 64, "condition": "Blood pressure"},
]

print(count_visits(patients))
```

Because the answer comes back from `return`, you can store it and use it again, not just print it once.

```python
result = count_visits(patients)
print("Diabetes visits:", result["Diabetes"])
```
